1. Given	a list,	remove the	element	at	index	number 4 and add	it at index position 2 (over-writing what is currently there) and also add it at the end of the list.

```python
list1 = [34, 56, 49, 4900, 249, 494]
```

````
list1 = [34, 56, 49, 4900, 249, 494]
# .pop remove by position
# .remove look at the number
item=list1.pop(4)

#this index position to and save it into a variable
list1[2]= item
print(list1)

#this put the variable at the end
(list1.append(item))
print(list1)
````
